```python
import os, glob
import xarray as xr
import s3fs
import imageio

import matplotlib.pyplot as plt
import numpy as np
import cmocean  # Oceanography colour maps
```

```python
#from utils import plot_swot_basemap
def plot_swot_basemap(ax, xlim, ylim):
    import cartopy.crs as ccrs
    import cartopy.feature as cfeature
    ax.set_extent([xlim[0], xlim[1], ylim[0], ylim[1]], crs=ccrs.PlateCarree())
    ax.coastlines(resolution="10m", color="black", linewidth=0.8)
    ax.add_feature(cfeature.LAND, facecolor="lightgray")
    gl = ax.gridlines(draw_labels=True, linewidth=0.5, color='gray', alpha=0.5)
    gl.top_labels = False
    gl.right_labels = False
    return ax
```

```python
def download_netcdf(files_on_s3, outdir="./DATA/GHRSST"):
    os.makedirs(outdir, exist_ok=True)
    local_files = []
    for key in files_on_s3:
        fname = os.path.basename(key)
        local_path = os.path.join(outdir, fname)
        if not os.path.exists(local_path):
            print(f"Downloading {key} -> {local_path}")
            fs.get(key, local_path)    # s3fs 下载
        else:
            print(f"Skip existing: {local_path}")
        local_files.append(local_path)
    return sorted(local_files)

local_nc_files = download_netcdf(aodn, outdir="./DATA/GHRSST")
print(f"local files: {len(local_nc_files)}")
```

```python
import os
import cartopy.crs as ccrs

def open_ds_any(ncfile: str):
    """本地路径用 xr.open_dataset；S3 键用 fs.open"""
    if ncfile.startswith("imos-data/") or ncfile.startswith("s3://"):
        with fs.open(ncfile) as f:
            return xr.open_dataset(f)
    else:
        return xr.open_dataset(ncfile)

def plot_scene(ncfile, timestr):
    print("Plot:", timestr, "|", ncfile)
    sst = open_ds_any(ncfile)

    # 裁剪经纬度范围（按需要改）
    x1, x2 = 111, 114
    y1, y2 = -25, -20
    mysst = sst.sel(lon=slice(x1, x2), lat=slice(y2, y1))

    # SST 变量名兜底
    for v in ["sea_surface_temperature", "analysed_sst", "sst"]:
        if v in mysst.variables:
            varname = v
            break
    else:
        raise KeyError(f"No SST variable found in {ncfile}")

    da = mysst[varname].squeeze()

    # 防御：像元至少 2x2 才能算梯度
    if (da.sizes.get("lat", 0) < 2) or (da.sizes.get("lon", 0) < 2):
        print(f"Skip (too small for gradient): {da.sizes}")
        return

    dx, dy = np.gradient(da.values)
    x = mysst['lon'].values
    y = mysst['lat'].values

    fig, ax = plt.subplots(figsize=(8, 6), subplot_kw={'projection': ccrs.PlateCarree()})

    da.plot(
        ax=ax,
        cmap=cmocean.cm.thermal,
        vmin=26+273.15, vmax=29+273.15,
        cbar_kwargs={"shrink": 0.3},
        transform=ccrs.PlateCarree()
    )

    ax.pcolormesh(
        x, y, np.abs(dx + 1j*dy),
        vmin=0, vmax=0.1, cmap='bone_r', alpha=0.5,
        transform=ccrs.PlateCarree()
    )

    plot_swot_basemap(ax, (x1, x2), (y1, y2))

    os.makedirs('../FIGURES', exist_ok=True)
    out_png = f'../FIGURES/GHRSST_1km_{timestr}.png'
    plt.tight_layout()
    plt.savefig(out_png, dpi=150)
    plt.close(fig)
    print("Saved:", out_png)


```

```python
# First import required libraries
import os
import glob
import json
import pandas as pd
import xarray as xr
import numpy as np
import s3fs

# Setup S3 filesystem connection
fs = s3fs.S3FileSystem(anon=True)

# First define the time period and get files
timestr = '202508'
aodn = fs.glob(f'imos-data/IMOS/SRS/SST/ghrsst/L3C-1d/ngt/snpp/2025/{timestr}*.nc')

# Download files
def download_and_process_sst(timestr, output_dir="../DATA/web"):
    """Download NC files and extract SST data"""
    # Download NC files first
    local_nc_files = download_netcdf(aodn, outdir="./DATA/GHRSST")
    print(f"Downloaded {len(local_nc_files)} files")
    
    # Process each file
    results = []
    for ncfile in local_nc_files:
        try:
            # Extract data
            sst = open_ds_any(ncfile)
            mysst = sst.sel(lon=slice(111, 114), lat=slice(-20, -25))
            
            # Find SST variable
            for v in ["sea_surface_temperature", "analysed_sst", "sst"]:
                if v in mysst.variables:
                    varname = v
                    break
            else:
                print(f"Skipping {ncfile}: No SST variable found")
                continue
                
            # Get data and convert to Celsius
            da = mysst[varname].squeeze()
            sst_celsius = da - 273.15
            
            # Create output files
            timestr = os.path.basename(ncfile)[0:8]
            
            # Save JSON
            json_data = {
                "timestamp": pd.Timestamp(da.time.values).isoformat(),
                "metadata": {
                    "region": "Ningaloo",
                    "lat_range": [float(da.lat.min()), float(da.lat.max())],
                    "lon_range": [float(da.lon.min()), float(da.lon.max())]
                },
                "data": {
                    "lats": da.lat.values.tolist(),
                    "lons": da.lon.values.tolist(),
                    "temperatures": sst_celsius.values.tolist(),
                    "units": "celsius"
                }
            }
            
            os.makedirs(output_dir, exist_ok=True)
            json_file = os.path.join(output_dir, f"sst_data_{timestr}.json")
            with open(json_file, 'w') as f:
                json.dump(json_data, f, indent=2)
                
            # Save CSV
            data_list = []
            for i in range(len(da.lat)):
                for j in range(len(da.lon)):
                    if not np.isnan(sst_celsius.values[i, j]):
                        data_list.append({
                            'latitude': float(da.lat[i]),
                            'longitude': float(da.lon[j]),
                            'temperature': float(sst_celsius.values[i, j])
                        })
            
            df = pd.DataFrame(data_list)
            csv_file = os.path.join(output_dir, f"sst_data_{timestr}.csv")
            df.to_csv(csv_file, index=False)
            
            print(f"Processed {timestr}: {len(data_list)} points")
            results.append((json_file, csv_file))
            
        except Exception as e:
            print(f"Error processing {ncfile}: {str(e)}")
            continue
            
    return results

# Run the processing
output_files = download_and_process_sst(timestr)
print(f"\nProcessed {len(output_files)} files")
```
